```python
import re
import unicodedata
import numpy as np

def strip_accents(s: str) -> str:
    return "".join(c for c in unicodedata.normalize("NFKD", s) if not unicodedata.combining(c))
```

```python
print(strip_accents("Féminin"))
print(strip_accents("Non précisé"))
```

```python
def normalize_key(s):
    if pd.isna(s):
        return None
    s = str(s)
    s = strip_accents(s).lower().strip()
    s = re.sub(r"[.\-_/]", " ", s)
    s = re.sub(r"\s+", " ", s)
    return s
```

```python
print(normalize_key("Pesticides et Pr. agricoles"))
print(normalize_key("PESTICIDES  ET   PRS. AGRICOLES"))
```

```python
def harmonize_by_keywords(series, rules, default_label="Non précisé", na_as_default=False):
    def mapper(x):
        if pd.isna(x):
            return default_label if na_as_default else np.nan
        key = normalize_key(x)
        if key == "" or key is None:
            return default_label if na_as_default else np.nan
        for label, keywords in rules:
            if any(kw in key for kw in keywords):
                return label
        return "Autre / non classé"
    return series.apply(mapper)
```

```python
import re

def extract_grade(x):
    if pd.isna(x):
        return np.nan
    m = re.search(r"\d", str(x))
    if m:
        g = int(m.group())
        if 0 <= g <= 5:
            return g
    return np.nan

df["pss_initial"] = df["gradation_initiale_brute"].apply(extract_grade)
df["pss_final"] = df["gradation_finale_brute"].apply(extract_grade)
df["cible_valide"] = df["pss_final"].isin([0, 1, 2, 3, 4])

print("=== pss_final ===")
print(df["pss_final"].value_counts(dropna=False).sort_index())
print()
print("=== pss_initial ===")
print(df["pss_initial"].value_counts(dropna=False).sort_index())
```
